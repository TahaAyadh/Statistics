```julia
age= rand(18:80,100)
wcc = round.(rand(Distributions.Normal(12,2),100),digits = 1)
crp = round.(Int,rand(Distributions.Chisq(4),100)).*10
treatment = rand(["A","B"],100)
result = rand(["Improved","Static","Worse"],100)
```
